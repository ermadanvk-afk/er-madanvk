Now making the vectors for each file, required for distance analysis of each of the file present.

```python
def get_word_frequency_vector(filename, bag_of_words):
    with open(filename, "r", encoding='utf-8') as f:
        text = f.read().lower()
        words = text.split(" ")
    freq_vector = [words.count(word) for word in bag_of_words]
    return freq_vector

# Example usage for polDoc1.txt
# matrix with row as file name and column as word in bag_of_words
matrix = []
for i in range(1,4):
    filename = f"polDoc{i}.txt"
    freq_vector = get_word_frequency_vector(filename, bag_of_words)
    matrix.append((filename, freq_vector))

for i in range(1,4):
    filename = f"sciDoc{i}.txt"
    freq_vector = get_word_frequency_vector(filename, bag_of_words)
    matrix.append((filename, freq_vector))

for i in range(1,4):
    filename = f"SDoc{i}.txt"
    freq_vector = get_word_frequency_vector(filename, bag_of_words)
    matrix.append((filename, freq_vector))

print(matrix)
```

Cosine Similarity and distance.

```python
import numpy as np
from numpy.linalg import norm               
def cosine_similarity(vec1, vec2):
    vec1 = np.array(vec1)
    vec2 = np.array(vec2)
    cos_sim = np.dot(vec1, vec2) / (norm(vec1) * norm(vec2))
    return cos_sim  
def cosine_distance(vec1, vec2):
    return 1 - cosine_similarity(vec1, vec2)
# Example: Cosine similarity and distance between polDoc1 and SciDoc1
cos_sim = cosine_similarity(vector_polDoc1, vector_SciDoc1)
cos_dist = cosine_distance(vector_polDoc1, vector_SciDoc1)  
print("Cosine Similarity between polDoc1 and SciDoc1:", cos_sim)    
print("Cosine Distance between polDoc1 and SciDoc1:", cos_dist)
```

K-L divergence 

```python
import numpy as np
from numpy.linalg import norm
def kl_divergence(p, q):
    p = np.array(p, dtype=np.float64)
    q = np.array(q, dtype=np.float64)
    p = p / np.sum(p)
    q = q / np.sum(q)
    p = np.where(p == 0, 1e-10, p)
    q = np.where(q == 0, 1e-10, q)
    return np.sum(p * np.log(p / q))
# Example: K-L divergence between polDoc1 and SciDoc1
kl_div = kl_divergence(vector_polDoc1, vector_SciDoc1)
print("K-L Divergence between polDoc1 and SciDoc1:", kl_div)
```
